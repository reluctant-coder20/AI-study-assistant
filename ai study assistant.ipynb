```python

!pip -q install gradio

import gradio as gr
import re

def study_assistant(text, task):
    if not text.strip():
        return "Please enter your study notes first."

    sentences = re.split(r'(?<=[.!?])\s+', text.strip())
    words = re.findall(r'\b[\w-]+\b', text.lower())

    if task == "Summarize notes":
        return "Summary:\n" + " ".join(sentences[:3])

    elif task == "Extract key points":
        return "Key points:\n" + "\n".join(
            f"• {s}" for s in sentences[:5]
        )

    elif task == "Generate questions":
        return "\n".join(
            f"{i+1}. Explain: {s}"
            for i, s in enumerate(sentences[:5])
        )

    else:
        return f"Word count: {len(words)}\nSentence count: {len(sentences)}"

demo = gr.Interface(
    fn=study_assistant,
    inputs=[
        gr.Textbox(lines=8, label="Paste your study notes"),
        gr.Radio(
            ["Summarize notes", "Extract key points",
             "Generate questions", "Analyze text"],
            label="Choose a task",
            value="Summarize notes"
        )
    ],
    outputs=gr.Textbox(lines=8, label="Result"),
    title="AI Study Assistant",
    description="A mini productivity tool for students"
)

demo.launch(share=True)
```
